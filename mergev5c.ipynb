```python
import os
import gc
import torch

from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
)

from peft import PeftModel


# =============================================================================
# V5C MERGE CONFIGURATION
# =============================================================================

BASE_MODEL = "./V5B_Final_Merged_Model"
V5C_LORA_PATH = "./V5C_Final/final_model/"
OUTPUT_PATH = "./V5C_Final_Merged_Model"


# =============================================================================
# CHECK PATHS
# =============================================================================

print("=" * 80)
print("V5C → FINAL MERGE")
print("=" * 80)

if not os.path.exists(BASE_MODEL):
    raise FileNotFoundError(
        f"V5B base model not found: {BASE_MODEL}"
    )

if not os.path.exists(V5C_LORA_PATH):
    raise FileNotFoundError(
        f"V5C LoRA adapter not found: {V5C_LORA_PATH}"
    )

os.makedirs(OUTPUT_PATH, exist_ok=True)


# =============================================================================
# LOAD TOKENIZER
# =============================================================================

print("\n" + "=" * 80)
print("1. LOADING TOKENIZER")
print("=" * 80)

tokenizer = AutoTokenizer.from_pretrained(
    BASE_MODEL,
    trust_remote_code=True,
)

print("Tokenizer loaded.")


# =============================================================================
# LOAD V5B BASE MODEL
# =============================================================================

print("\n" + "=" * 80)
print("2. LOADING V5B BASE MODEL")
print("=" * 80)

print("Loading on CPU...")
print("This is intentional.")
print("RTX 3060 is NOT used for the merge.")

base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,

    # Important: load in FP16 for correct LoRA merging
    torch_dtype=torch.float16,

    # IMPORTANT:
    # Do NOT use device_map='auto' for this merge.
    device_map="cpu",

    low_cpu_mem_usage=True,

    trust_remote_code=True,
)

print("V5B base model loaded successfully.")


# =============================================================================
# LOAD V5C LORA
# =============================================================================

print("\n" + "=" * 80)
print("3. LOADING V5C LORA ADAPTER")
print("=" * 80)

model = PeftModel.from_pretrained(
    base_model,
    V5C_LORA_PATH,
)

print("V5C LoRA loaded successfully.")


# =============================================================================
# MERGE
# =============================================================================

print("\n" + "=" * 80)
print("4. MERGING V5C LORA INTO V5B")
print("=" * 80)

print("Merging...")
print("Please wait.")

merged_model = model.merge_and_unload()

print("V5C LoRA merged successfully.")


# =============================================================================
# SAVE MERGED MODEL
# =============================================================================

print("\n" + "=" * 80)
print("5. SAVING V5C FINAL MERGED MODEL")
print("=" * 80)

print("Output:")
print(OUTPUT_PATH)

merged_model.save_pretrained(
    OUTPUT_PATH,
    safe_serialization=True,
    max_shard_size="4GB",
)

tokenizer.save_pretrained(
    OUTPUT_PATH
)

print("Model saved successfully.")


# =============================================================================
# CLEAN MEMORY
# =============================================================================

del model
del base_model
del merged_model

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()


# =============================================================================
# COMPLETE
# =============================================================================

print("\n" + "=" * 80)
print("✅ V5C MERGE COMPLETE")
print("=" * 80)

print()
print("Final model:")
print(OUTPUT_PATH)

print()
print("Pipeline:")
print("V5A → V5A Merged")
print("V5B → V5B Merged")
print("V5C → V5C Final Merged")

print("=" * 80)
```
