```python
import torch

from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import PeftModel


BASE_MODEL = "./V5A_Final_Merged_Model"

# Your trained V5B adapter
V5B_LORA_PATH = "./v5b_full_checkpoints/final_model/"

# Output merged model
OUTPUT_PATH = "./V5B_Final_Merged_Model"

```

```python

# ==============================================================================
# LOAD TOKENIZER
# ==============================================================================

print("=" * 70)
print("LOADING TOKENIZER")
print("=" * 70)

tokenizer = AutoTokenizer.from_pretrained(
    BASE_MODEL,
    trust_remote_code=True
)


# ==============================================================================
# LOAD BASE MODEL
# ==============================================================================

print("=" * 70)
print("LOADING BASE MODEL")
print("=" * 70)

base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    torch_dtype=torch.float16,
    device_map="auto",
    low_cpu_mem_usage=True,
    trust_remote_code=True,
)


# ==============================================================================
# LOAD V5B LoRA
# ==============================================================================

print("=" * 70)
print("LOADING V5B LoRA")
print("=" * 70)

model = PeftModel.from_pretrained(
    base_model,
    V5B_LORA_PATH,
)


# ==============================================================================
# MERGE V5B LoRA
# ==============================================================================

print("=" * 70)
print("MERGING V5B LoRA")
print("=" * 70)

merged_model = model.merge_and_unload()


# ==============================================================================
# SAVE V5B FINAL MERGED MODEL
# ==============================================================================

print("=" * 70)
print("SAVING V5B FINAL MERGED MODEL")
print("=" * 70)

merged_model.save_pretrained(
    OUTPUT_PATH,
    safe_serialization=True,
)

tokenizer.save_pretrained(
    OUTPUT_PATH
)


# ==============================================================================
# COMPLETE
# ==============================================================================

print()
print("=" * 70)
print("V5B MERGE COMPLETE")
print("=" * 70)
print("Final model saved to:")
print(OUTPUT_PATH)
print("=" * 70)
```
